```python
import matplotlib.pyplot as plt
days = ["mon","tue","wed","thur","fri"]
scores = [67,78,90,45,56]
plt.plot(days,scores,color = "crimson",marker ="x",linestyle ="dashdot",linewidth = 5)
plt.title("MY FIRST GRAPH")
plt.xlabel("DAYS OF THE WEEK")
plt.ylabel("TEST SCORES")
plt.grid()
plt.ylim(0,100)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.bar(days,scores,color = "purple")
plt.title("My first Bar CHART")
plt.xlabel("Days of the week")
plt.ylabel("SCORE")
plt.ylim(0,100)
plt.show()
```
